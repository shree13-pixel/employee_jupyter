## Database Design
Table: people
- emp_id (INT, Primary Key)
- name (VARCHAR)
- department (VARCHAR)
- salary (INT)


```python
def add_employee():
    emp_id = int(input("Enter Employee ID: "))
    name = input("Enter Name: ")
    department = input("Enter Department: ")
    salary = int(input("Enter Salary: "))

    df = pd.DataFrame(
        [[emp_id, name, department, salary]],
        columns=["emp_id", "name", "department", "salary"]
    )

    df.to_sql("employees", conn, if_exists="append", index=False)
    print("✅ Employee added successfully")
```

```python
def view_employee():
    df = pd.read_sql("SELECT * FROM people", conn)
    display(df)
```

```python
def update_employee():
    emp_id = int(input("Enter Employee ID to update: "))
    new_salary = int(input("Enter new salary: "))

    conn.execute(
        "UPDATE employees SET salary = ? WHERE emp_id = ?",
        (new_salary, emp_id)
    )
    conn.commit()
    print("✏️ Employee updated successfully")
```

```python
def delete_employee():
    emp_id = int(input("Enter Employee ID to delete: "))

    conn.execute(
        "DELETE FROM employees WHERE emp_id = ?",
        (emp_id,)
    )
    conn.commit()
    print("🗑️ Employee deleted successfully")
```

```python
import pandas as pd
def employee_graph():
    df = pd.read_sql("SELECT * FROM people", conn)

    if df.empty:
        print("No data to display graph")
        return

    df['job_title'].value_counts().plot(kind='line')
    plt.title("Employees per Department")
    plt.xlabel("Department")
    plt.ylabel("Number of Employees")
    plt.show()
```
